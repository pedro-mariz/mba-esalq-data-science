# Estrutura básica de tensores

Tensores são a estrutura de dados fundamental do deep learning. Aqui
percorremos os ranks 0, 1, 2 e 3, observando sempre **valor**, **rank**,
**shape** e **tipo**.

**Atenção**: *Material complementar para estudo individual. Não se preocupe em compreender os detalhes neste momento. Trabalharemos isso ao longo das aulas.*

## Importação de bibliotecas

```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
from absl import logging

tf.get_logger().setLevel('ERROR')
logging.set_verbosity(logging.ERROR)
```

## Funções auxiliares

Execute esta célula uma vez, no início da sessão. Ela define as funções de
apoio usadas ao longo do notebook — não é preciso importar nada além do que
já foi importado acima.

```python
def plot_images(images, scale=1.0):
    """
    Plota uma lista de imagens.
    """
    num_images = len(images)
    fig, axes = plt.subplots(1, num_images, figsize=(scale * num_images, scale))
    if num_images == 1:
        axes = [axes]
    for ax, img in zip(axes, images):
        ax.imshow(img, cmap='gray')
        ax.axis('off')
    plt.tight_layout()
    plt.show()
```

## Tensor de rank-0 (ESCALAR)

```python
t1 = np.array(42.)

print(f'valor: {t1}')
print(f'rank: {t1.ndim}')
print(f'shape: {t1.shape}')
print(f'tipo: {t1.dtype}')
```

## Tensores de rank-1 (VETOR)

```python
t1 = np.array([1, 2, 3, 4])

print(f'valor: {t1}')
print(f'rank: {t1.ndim}')
print(f'shape: {t1.shape}')
print(f'tipo: {t1.dtype}')
```

```python
t2 = np.random.randint(low=0, high=256, size=(10,))

print(f'valor: {t2}')
print(f'rank: {t2.ndim}')
print(f'shape: {t2.shape}')
print(f'tipo: {t2.dtype}')
```

## Tensores de rank-2 (MATRIZ)

```python
t1 = np.array([
    [1, 2, 3, 4],
    [10, 20, 30, 40]
])

print(f'valor: \n{t1}')
print(f'rank: {t1.ndim}')
print(f'shape: {t1.shape}')
print(f'tipo: {t1.dtype}')
```

```python
t2 = np.random.randint(low=0, high=256, size=(3, 5))

print(f'valor: \n{t2}')
print(f'rank: {t2.ndim}')
print(f'shape: {t2.shape}')
print(f'tipo: {t2.dtype}')
```

## Tensores de rank-3 (HIPERDIMENSIONAIS)

```python
t1 = np.array([
    [
        [1, 2, 3, 4],
        [10, 20, 30, 40]
    ],
    [
        [15, 30, 60, 90],
        [11, 12, 13, 14]
    ]
])

print(f'valor: \n{t1}')
print(f'rank: {t1.ndim}')
print(f'shape: {t1.shape}')
print(f'tipo: {t1.dtype}')
```

```python
t2 = np.random.randint(low=0, high=256, size=(2, 5, 3))

print(f'valor: \n{t2}')
print(f'rank: {t2.ndim}')
print(f'shape: {t2.shape}')
print(f'tipo: {t2.dtype}')
```

## Tensores como imagens

A simples adição de um rank em um tensor pode modificar completamente
sua interpretação.

```python
plot_images(t2, scale=4)
```

```python
plot_images([t2], scale=4)
```
