## Análise de Sentimentos

A ideia é simples: temos uma lista de palavras "positivas" e outra de "negativas".
Para cada frase contamos quantas palavras de cada tipo aparecem e decidimos o sentimento
pelo **saldo** (quem tiver mais, vence; empate = neutro).

### 1) O léxico: nossas listas de palavras

Tudo começa com duas listas simples: palavras que soam **positivas** e palavras que soam **negativas**.

````python
palavras_positivas = ["bom", "boa", "ótimo", "excelente", "gostei", "adorei", "feliz"]
palavras_negativas = ["ruim", "péssimo", "horrível", "odiei", "triste", "chato"]
````

### 2) As frases que vamos analisar

Cinco frases de exemplo. Repare que a última não tem palavra positiva nem negativa.

````python
frases = [
    "O produto é ótimo e eu adorei a entrega",
    "Que serviço péssimo, odiei o atendimento",
    "O filme foi bom mas o final foi ruim",
    "Estou muito feliz com a compra",
    "O relatório foi entregue",
]
````

### 3) Pré processamento

Antes do loop, vamos processar **só a primeira frase**. O truque é deixar tudo em
minúsculo e quebrar o texto em uma lista de palavras.

````python
frase = frases[0]

palavras = frase.lower().split()
palavras
````

### 4) Contando as palavras positivas e negativas

Percorremos palavra por palavra e somamos um ponto para cada lado.

````python
positivos = 0
negativos = 0

for palavra in palavras:
    if palavra in palavras_positivas:
        positivos += 1
    elif palavra in palavras_negativas:
        negativos += 1

print("positivos:", positivos, "| negativos:", negativos)
````

### 5) Decidindo o sentimento pelo saldo

Quem tiver mais pontos vence. Empate = neutro.

````python
if positivos > negativos:
    sentimento = "Positivo"
elif negativos > positivos:
    sentimento = "Negativo"
else:
    sentimento = "Neutro"

print(sentimento)
````

### 6) 5 frases
Agora que a lógica funciona para uma frase, é só colocá-la dentro de um `for`
que passa por todas as frases.

````python
for frase in frases:
    palavras = frase.lower().split()

    positivos = 0
    negativos = 0
    for palavra in palavras:
        if palavra in palavras_positivas:
            positivos += 1
        elif palavra in palavras_negativas:
            negativos += 1

    if positivos > negativos:
        sentimento = "Positivo"
    elif negativos > positivos:
        sentimento = "Negativo"
    else:
        sentimento = "Neutro"

    print(f"{sentimento:8} | +{positivos} / -{negativos} | {frase}")
````

### 7) Criando uma função

````python
def classificar_sentimento(frase):
    # Léxico simples de palavras em português
    palavras_positivas = [
        "bom", "boa", "ótimo", "excelente", "maravilhoso", "gostei", "gosto",
        "feliz", "amor", "adoro", "sensacional", "perfeito", "parabéns", "sucesso"
    ]
    palavras_negativas = [
        "ruim", "péssimo", "horrível", "odiei", "odeio", "triste", "pena",
        "chato", "difícil", "problema", "erro", "fracasso", "mal", "pior"
    ]

    # 1. Pré-processamento: minúsculas + remoção de pontuação
    frase_limpa = frase.lower()
    for pontuacao in [".", ",", "!", "?", ";", ":"]:
        frase_limpa = frase_limpa.replace(pontuacao, "")
    palavras = frase_limpa.split()

    # 2. Contagem de palavras
    positivos = 0
    negativos = 0
    for palavra in palavras:
        if palavra in palavras_positivas:
            positivos += 1
        elif palavra in palavras_negativas:
            negativos += 1

    # 3. Classificação pelo saldo
    if positivos > negativos:
        sentimento = "Positivo"
    elif negativos > positivos:
        sentimento = "Negativo"
    else:
        sentimento = "Neutro"

    # 4. Mostra a saída no mesmo formato do passo anterior
    print(f"{sentimento:8} | +{positivos} / -{negativos} | {frase}")
````

````python
classificar_sentimento("O atendimento foi excelente, adorei a experiência!")
````

````python
classificar_sentimento("O atendimento foi excelente, adorei a experiência!")
````

# Como essa lib classifica os sentimentos?

## Fluxo

```text
Texto
↓
Tokens
↓
Embeddings
↓
Transformer
↓
Logits
↓
Softmax
↓
Classe + Confiança
```

## 1. Tokenização

O texto é dividido em **tokens** (palavras ou partes de palavras).

Exemplo:

```text
"Gostei muito!"
↓
["Gostei", "muito", "!"]
```

---

## 2. Embeddings

Cada token é convertido em um **vetor numérico** que representa seu significado.

---

## 3. Transformer

O modelo analisa o **contexto** entre as palavras.

Exemplo:

```text
"Não gostei."
```

Mesmo que **"gostei"** seja uma palavra positiva, o modelo percebe que **"Não"** muda o significado da frase para negativo.

---

## 4. Logits

O modelo calcula uma pontuação para cada classe.

| Classe | Logit |
|---------|------:|
| Positivo | 8.3 |
| Neutro | 1.4 |
| Negativo | -2.8 |

Esses valores **não são probabilidades**.

---

## 5. Softmax

A função **Softmax** transforma os logits em probabilidades.

| Classe | Probabilidade |
|---------|--------------:|
| Positivo | 99.7% |
| Neutro | 0.2% |
| Negativo | 0.1% |

A classe com maior probabilidade é escolhida.

---

## Resultado

```python
{
    "label": "positive",
    "score": 0.997
}
```

- **label:** sentimento previsto pelo modelo.
- **score:** confiança do modelo nessa previsão (99,7% neste exemplo).
